```python

import pandas as pd
import numpy as np

# Create a working copy of the dataset
data = df.copy()

# Convert question marks into missing values
data = data.replace("?", np.nan)

# Check and remove duplicate rows
duplicates_before = data.duplicated().sum()
data = data.drop_duplicates().reset_index(drop=True)

# Create our prediction target
# 1 = readmitted within 30 days
# 0 = readmitted after 30 days or not readmitted
data["target"] = (data["readmitted"] == "<30").astype(int)

# Display the results
print("Original dataset shape:", df.shape)
print("Duplicate rows found:", duplicates_before)
print("Cleaned dataset shape:", data.shape)

print("\nTarget counts:")
print(data["target"].value_counts())

print("\nTarget percentages:")
print((data["target"].value_counts(normalize=True) * 100).round(2))

print("\nColumns with the most missing values:")
print(data.isna().sum().sort_values(ascending=False).head(10))
```

```python

# Columns that should not be used as model inputs
columns_to_drop = [
    "encounter_id",
    "patient_nbr",
    "readmitted",
    "target"
]

# Separate input features (X) and target label (y)
X = data.drop(columns=columns_to_drop)
y = data["target"]

print("Number of input features:", X.shape[1])
print("Number of rows:", X.shape[0])

print("\nTarget distribution:")
print(y.value_counts())

print("\nData types of input features:")
print(X.dtypes.value_counts())

print("\nFirst five rows of input features:")
display(X.head())
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Identify categorical and numerical columns
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))

# Preprocess numerical features
numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocess categorical features
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=True
    ))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer([
    ("num", numerical_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("\nPreprocessing pipeline created successfully!")
```

```python

from sklearn.model_selection import GroupShuffleSplit

# Patient IDs are used only to keep each patient in one split
groups = data["patient_nbr"]

# Split patients into training and test groups
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

train_patients = set(groups.iloc[train_idx])
test_patients = set(groups.iloc[test_idx])

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))

print("Training patients:", len(train_patients))
print("Test patients:", len(test_patients))

print("Patients appearing in both sets:",
      len(train_patients.intersection(test_patients)))

print("\nTraining target percentages:")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTest target percentages:")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Create a complete pipeline:
# preprocessing followed by Logistic Regression
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train, y_train)

# Predict test-set labels and probabilities
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

# Evaluate the model
print("Logistic Regression Results")
print("----------------------------")
print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, y_prob), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["No readmission within 30 days", "Readmission within 30 days"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Build the Random Forest pipeline
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=150,
        max_depth=20,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model
rf_model.fit(X_train, y_train)

# Predict on the same held-out test set
rf_pred = rf_model.predict(X_test)
rf_prob = rf_model.predict_proba(X_test)[:, 1]

# Evaluate
print("Random Forest Results")
print("----------------------")
print("Accuracy:", round(accuracy_score(y_test, rf_pred), 4))
print("Precision:", round(precision_score(y_test, rf_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, rf_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, rf_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, rf_prob), 4))
```

```python

from sklearn.model_selection import GroupShuffleSplit

# Split the existing training data by patient
train_groups = data.loc[X_train.index, "patient_nbr"]

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=123
)

fit_idx, val_idx = next(
    splitter.split(X_train, y_train, groups=train_groups)
)

X_fit = X_train.iloc[fit_idx]
y_fit = y_train.iloc[fit_idx]

X_val = X_train.iloc[val_idx]
y_val = y_train.iloc[val_idx]

fit_patients = set(train_groups.iloc[fit_idx])
val_patients = set(train_groups.iloc[val_idx])

print("Model-training rows:", len(X_fit))
print("Validation rows:", len(X_val))
print("Patients shared between training and validation:",
      len(fit_patients.intersection(val_patients)))
```

```python

import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.metrics import (
    precision_score,
    recall_score,
    fbeta_score
)

# Train a fresh Logistic Regression model on the fit subset
lr_tuned = clone(logistic_model)
lr_tuned.fit(X_fit, y_fit)

# Predict probabilities on validation data
val_prob = lr_tuned.predict_proba(X_val)[:, 1]

# Compare candidate thresholds
thresholds = np.arange(0.10, 0.71, 0.05)
rows = []

for threshold in thresholds:
    val_pred = (val_prob >= threshold).astype(int)

    rows.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(
            y_val, val_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_val, val_pred, zero_division=0
        ),
        "F2-score": fbeta_score(
            y_val, val_pred, beta=2, zero_division=0
        )
    })

threshold_results = pd.DataFrame(rows)

print(threshold_results.round(4).to_string(index=False))

# Select the threshold with the highest validation F2-score
best_row = threshold_results.loc[
    threshold_results["F2-score"].idxmax()
]

best_threshold = float(best_row["Threshold"])

print("\nBest validation threshold:", best_threshold)
print("Best validation F2-score:", round(best_row["F2-score"], 4))
print("Validation precision:", round(best_row["Precision"], 4))
print("Validation recall:", round(best_row["Recall"], 4))
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Apply the selected validation threshold to the test probabilities
test_prob_tuned = lr_tuned.predict_proba(X_test)[:, 1]
test_pred_tuned = (
    test_prob_tuned >= best_threshold
).astype(int)

print("Tuned Logistic Regression — Test Results")
print("-----------------------------------------")
print("Threshold:", best_threshold)
print("Accuracy:", round(accuracy_score(y_test, test_pred_tuned), 4))
print("Precision:", round(precision_score(y_test, test_pred_tuned, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, test_pred_tuned, zero_division=0), 4))
print("F1-score:", round(f1_score(y_test, test_pred_tuned, zero_division=0), 4))
print("F2-score:", round(fbeta_score(y_test, test_pred_tuned, beta=2, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, test_prob_tuned), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, test_pred_tuned))

print("\nClassification Report:")
print(classification_report(
    y_test,
    test_pred_tuned,
    target_names=[
        "No readmission within 30 days",
        "Readmission within 30 days"
    ],
    zero_division=0
))
```

```python

import shap
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt

# Sample held-out test encounters
X_shap = X_test.sample(
    n=min(200, len(X_test)),
    random_state=42
)

# Apply the preprocessing already fitted by the Random Forest
X_shap_transformed = preprocessor_rf.transform(X_shap)

# SHAP plotting requires a dense array for this workflow
if sp.issparse(X_shap_transformed):
    X_shap_transformed = X_shap_transformed.toarray()

# Create the SHAP explainer for the trained Random Forest
explainer = shap.TreeExplainer(rf_classifier)
shap_values = explainer.shap_values(X_shap_transformed)

# Handle different SHAP output formats for binary classification
if isinstance(shap_values, list):
    positive_shap_values = shap_values[1]
elif np.asarray(shap_values).ndim == 3:
    positive_shap_values = np.asarray(shap_values)[:, :, 1]
else:
    positive_shap_values = np.asarray(shap_values)

# Summary plot for the readmission class (target = 1)
shap.summary_plot(
    positive_shap_values,
    X_shap_transformed,
    feature_names=feature_names,
    max_display=15,
    show=False
)

plt.title("SHAP Summary: Factors Influencing Readmission Predictions")
plt.tight_layout()
plt.savefig("shap_summary_readmission.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
